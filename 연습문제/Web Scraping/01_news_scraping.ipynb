### 웹스크래핑 연습문제 1-1

```python
from bs4 import BeautifulSoup
import requests

# 1. 다음 경제 뉴스 URL 지정 및 출력
url = "https://news.daum.net/economy"
print(url)

# 브라우저 요청 헤더 설정
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML,"
        " like Gecko) Chrome/120.0.0.0 Safari/537.36"
    )
}

# 2. HTTP GET 요청 및 Response 객체 정보 출력
res = requests.get(url, headers=headers)
print(type(res))
print(res.status_code)

if res.ok:
    # 한글 깨짐 방지를 위해 res.content 사용
    soup = BeautifulSoup(res.content, "html.parser")

    # 3. 변경사항: 다음 기사 상세 페이지 URL 패턴(v.daum.net/v/)을 갖는 모든 <a> 태그 추출
    a_tags = soup.select("a[href*='v.daum.net/v/']")
    print(type(a_tags), len(a_tags))

    seen_links = set()

    # 4. 기사 링크와 제목 출력
    for a in a_tags:
        link = a.get("href", "")
        title = " ".join(a.get_text().split())

        # 텍스트(제목)가 존재하고, 중복되지 않은 기사만 출력
        if title and link not in seen_links:
            seen_links.add(link)
            print(link)
            print(title)
else:
    print(f"요청 실패 (상태 코드: {res.status_code})")
```

### 웹스크래핑 연습문제 1-2

```python
import requests
from bs4 import BeautifulSoup

# 다음 뉴스 섹션 사전
section_dict = {'기후/환경': 'climate','사회': 'society','경제': 'economy','정치': 'politics','국제': 'world','문화': 'culture','생활': 'life','IT/과학': 'tech','인물': 'people'}

def print_news(section_name):
    # section_dict에서 섹션 영문 코드 가져오기 (기본값 'economy')
    section_code = section_dict.get(section_name, 'economy')
    url = f'https://news.daum.net/{section_code}'
    
    # 출력 헤더
    print(f"======> {url} {section_name} 뉴스 <======")

    # 요청 헤더 설정 : 브라우저 정보
    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36'
    }

    # requests의 get() 함수 호출하기
    res = requests.get(url, headers=req_header)
    # 한글 깨짐 방지
    res.encoding = 'utf-8'

    # 응답(response)이 OK 이면
    if res.ok:
        # 응답(response)에서 text 추출
        html = res.text
        # BeautifulSoup 객체 생성
        soup = BeautifulSoup(html, 'html.parser')

        # 다음 뉴스 기사 상세페이지 경로(/v/ 또는 v.daum.net/v/)를 포함하는 a 태그 선택
        a_tags = soup.select("a[href*='/v/']")

        # 중복 출력 방지용 set
        seen_links = set()

        # <a> 태그 리스트 순회하기
        for a_tag in a_tags:
            title = a_tag.text.strip()  # <a>title</a>
            link = a_tag.get('href', '')  # <a href="link">

            # 상대 경로 표준화 (https: 붙이기)
            if link.startswith('//'):
                link = 'https:' + link
            elif link.startswith('/'):
                link = 'https://news.daum.net' + link

            # 제목이 존재하고 중복되지 않은 기사만 출력
            if title and link not in seen_links:
                seen_links.add(link)
                print(link)
                print(title)

    # 응답(response)이 Error 이면 status code 출력
    else:
        print(f'Error Code = {res.status_code}')

# 함수 호출 테스트
print_news('경제')
```
