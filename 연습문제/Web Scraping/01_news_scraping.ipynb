### 웹스크래핑 연습문제 1-1

```python
import requests
from bs4 import BeautifulSoup

# 다음 경제 뉴스 URL
url = 'https://news.daum.net/economy'
print(url)

# 요청 헤더 설정 : 브라우저 정보
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36'
}

# requests 의 get() 함수 호출하기 
res = requests.get(url, headers=req_header)
# 한글 깨짐 방지
res.encoding = 'utf-8'

print(type(res))
print(res.status_code)

# 응답(response)이 OK 이면
if res.ok:    
    # 응답 (response)에서 text 추출
    html = res.text
    # BeautifulSoup 객체 생성  
    soup = BeautifulSoup(html, 'html.parser')
    
    # CSS 선택자 (다음 뉴스 경제 섹션의 기사 제목/링크 a 태그 선택)
    a_tags = soup.select("ul.list_newsissue a.link_txt, div.box_g a.link_txt, a[href*='v.daum.net/v/']")
    
    # 중복 및 유효한 제목(텍스트) 필터링
    filtered_tags = []
    seen_links = set()
    
    for a in a_tags:
        title = a.text.strip()
        link = a.get('href', '')
        
        # 상대 경로 처리 (https: 붙이기)
        if link.startswith('//'):
            link = 'https:' + link
        elif link.startswith('/'):
            link = 'https://news.daum.net' + link
            
        # 텍스트가 존재하고 중복되지 않은 링크만 정제
        if title and link not in seen_links:
            seen_links.add(link)
            filtered_tags.append((link, title))
            
    print(type(a_tags), len(filtered_tags))
    
    # <a> 태그 순회하며 출력    
    for link, title in filtered_tags:    
        print(link)
        print(title)

# 응답(response)이 Error 이면 status code 출력    
else:
    print(f'Error Code = {res.status_code}')
```

### 웹스크래핑 연습문제 1-2

```python
import requests
from bs4 import BeautifulSoup

# Daum 뉴스 섹션 사전 (한글명 : 영문 코드)
section_dict = {'기후/환경': 'climate','사회': 'society','경제': 'economy','정치': 'politics','국제': 'world','문화': 'culture','생활': 'life','IT/과학': 'tech','인물': 'people'}

def print_news(section_name):
    # section_dict에서 섹션 영문 코드 가져오기 (기본값 'economy')
    section_code = section_dict.get(section_name, 'economy')
    url = f'https://news.daum.net/{section_code}'
    print(f'======> {url} {section_name} 뉴스 <======')

    # 요청 헤더 설정 : 브라우저 정보
    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36'
    }

    # requests 의 get() 함수 호출하기 
    res = requests.get(url, headers=req_header)
    # 한글 깨짐 방지
    res.encoding = 'utf-8'

    # 응답(response)이 OK 이면
    if res.ok:    
        # 응답 (response)에서 text 추출
        html = res.text
        # BeautifulSoup 객체 생성  
        soup = BeautifulSoup(html, 'html.parser')
        
        # CSS 선택자 (다음 뉴스 기사 본문 링크 선택)
        a_tags = soup.select("a[href*='/v/']")
        
        # 중복 출력 방지용 집합
        seen_links = set()
        
        # <a> 태그 리스트 순회하기    
        for a_tag in a_tags:    
            title = a_tag.text.strip()  # <a>title</a>
            link = a_tag.get('href', '')  # <a href="link">
            
            # 상대 경로 표준화 (https: 붙이기)
            if link.startswith('//'):
                link = 'https:' + link
            elif link.startswith('/'):
                link = 'https://news.daum.net' + link
                
            # 제목이 존재하고 중복되지 않은 기사만 출력
            if title and link not in seen_links:
                seen_links.add(link)
                print(link)
                print(title)

    # 응답(response)이 Error 이면 status code 출력    
    else:
        print(f'Error Code = {res.status_code}')

# 함수 호출 테스트
print_news('경제')
```
