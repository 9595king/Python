### 1. 지역별 총인구수 Top5

```python
import pandas as pd

# 1. 파일 불러오기 (한글 인코딩 대응)
try:
    df = pd.read_csv("../data/인구현황.csv", encoding="utf-8")
except UnicodeDecodeError:
    df = pd.read_csv("../data/인구현황.csv", encoding="cp949")

# 2. 조건 추출
top5 = df[df["행정기관"] != "전국"].sort_values(by="총인구수", ascending=False).head(5)[["행정기관", "총인구수"]]

# 3. 출력
print("1. 지역별 총인구수 Top5")
print(top5)

df.sort_values(by='총인구수', ascending=False)[['행정기관', '총인구수']].head(5)
```

### 1-1. 지역별 총인구수 plot

### 3-1. 남녀 비율 분석 Plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 (Windows: Malgun Gothic / Mac: AppleGothic)
plt.rc('font', family='Malgun Gothic') 
# plt.rc('font', family='AppleGothic') # Mac 사용자일 경우 주석 해제

plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호 깨짐 방지

# 데이터 불러오기
df = pd.read_csv('../data/인구현황.csv')

# 3-1. 남여 비율 분석 Plot
plt.figure(figsize=(10, 5))

# 내림차순 정렬 데이터 생성
df_gender = df.sort_values(by='남여 비율', ascending=False)

sns.barplot(data=df_gender, x='행정기관', y='남여 비율',hue="행정기관")

# 기준이 되는 빨간 점선 추가 (axhline)
plt.axhline(y=1.0, color='red', linestyle='--', label='남여 1:1 비율')

plt.title('지역별 남녀 비율')
plt.xticks(rotation=45) # x축 라벨 45도 회전
plt.legend() # 범례 표시
plt.tight_layout()
plt.show()
```

### 5. 세대당 인구 평균보다 높은 지역

```python
# 5. 세대당 인구 평균보다 높은 지역 (loc 조건 검색 사용)
mean_hh_val = data['세대당 인구'].mean()
print(f"전국 평균 세대당 인구수: {mean_hh_val:.2f}")

above_mean_df = data.loc[data['세대당 인구'] > mean_hh_val, ['행정기관', '세대당 인구']]\
                    .sort_values(by='세대당 인구', ascending=False)\
                    .reset_index(drop=True)
print("5. 세대당 인구 평균보다 높은 지역")
display(above_mean_df)
```
